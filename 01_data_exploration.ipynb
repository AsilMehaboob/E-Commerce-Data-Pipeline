```python
base_path = "/Volumes/workspace/default/ecommerce_raw/"

customers_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv(base_path + "olist_customers_dataset.csv")

order_items_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv(base_path + "olist_order_items_dataset.csv")

payments_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv(base_path + "olist_order_payments_dataset.csv")

reviews_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv(base_path + "olist_order_reviews_dataset.csv")

products_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv(base_path + "olist_products_dataset.csv")

sellers_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv(base_path + "olist_sellers_dataset.csv")

geolocation_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv(base_path + "olist_geolocation_dataset.csv")

category_translation_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv(base_path + "product_category_name_translation.csv")
```

```python
datasets = {
    "customers": customers_df,
    "orders": orders_df,
    "order_items": order_items_df,
    "payments": payments_df,
    "reviews": reviews_df,
    "products": products_df,
    "sellers": sellers_df,
    "geolocation": geolocation_df,
    "category_translation": category_translation_df
}

for name, df in datasets.items():
    print(f"\n{name}")
    print("-" * len(name))
    print("Rows:", df.count())
    print("Columns:", df.columns)
```

```python
print("Customers:", customers_df.select("customer_id").distinct().count())
print("Orders:", orders_df.select("order_id").distinct().count())
print("Products:", products_df.select("product_id").distinct().count())
print("Sellers:", sellers_df.select("seller_id").distinct().count())
```

```python
orders_without_customer = orders_df.join(
    customers_df,
    on="customer_id",
    how="left_anti"
)

print(
    "Orders with no matching customer:",
    orders_without_customer.count()
)
```

```python
items_without_order = order_items_df.join(
    orders_df,
    on="order_id",
    how="left_anti"
)

print(
    "Order items with no matching order:",
    items_without_order.count()
)
```

```python
items_without_product = order_items_df.join(
    products_df,
    on="product_id",
    how="left_anti"
)

print(
    "Order items with no matching product:",
    items_without_product.count()
)
```

```python
items_without_seller = order_items_df.join(
    sellers_df,
    on="seller_id",
    how="left_anti"
)

print(
    "Order items with no matching seller:",
    items_without_seller.count()
)
```

```python
print("Customer rows:", customers_df.count())
print("Unique customer_id:", customers_df.select("customer_id").distinct().count())
print()

print("Product rows:", products_df.count())
print("Unique product_id:", products_df.select("product_id").distinct().count())
print()

print("Seller rows:", sellers_df.count())
print("Unique seller_id:", sellers_df.select("seller_id").distinct().count())
```
